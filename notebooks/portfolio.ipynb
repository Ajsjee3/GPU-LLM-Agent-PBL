# 프로젝트 소개 초안 작성
05 Web MVP의 Qwen3-8B 생성 함수를 재사용합니다. 같은 GPU 환경에서 새 커널을 열고 프로젝트 이름, 사용 기술, 직접 구현한 기능을 입력하세요. 생성된 소개와 배운 점은 실제 경험에 맞는지 검토하세요.

```python
import sys
from pathlib import Path

root = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / "05_agent_web_mvp/app/llm.py").is_file()), None)
if root is None:
    raise FileNotFoundError("저장소를 복제한 뒤 notebooks 폴더에서 실행하세요.")
sys.path.insert(0, str(root / "05_agent_web_mvp"))

from app.llm import generate_llm

# ① 프로젝트 정보 입력
project = input("📌 프로젝트 이름: ")
skills = input("💻 사용한 기술: ")
role = input("👨‍💻 내가 구현한 기능: ")

# ② AI에게 포트폴리오 작성 요청
prompt = f"""
다음 정보를 바탕으로 대학생 취업 포트폴리오를 작성해줘.

프로젝트 이름: {project}
사용 기술: {skills}
내가 구현한 기능: {role}

다음 4개 항목으로 작성해줘.
1. 프로젝트 소개
2. 사용 기술
3. 주요 구현 기능
4. 프로젝트를 통해 배운 점

GitHub README에 적합한 Markdown 형식으로 작성해줘.
제공하지 않은 성과나 경험은 만들어내지 마.
"""

# ③ Qwen3-8B 실행
result = generate_llm(
    [{"role": "user", "content": prompt}],
    max_new_tokens=500
)

print("\n📄 나의 AI 포트폴리오")
print(result)
```
